## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My chosen lane is Lane 2: Refresh / Content Opportunity Scoring.

The ML task type is scoring/ranking with a classification proxy.

This means I want to score or rank content pages so a content team can decide which pages should be reviewed first for refresh, improvement, protection, pruning, or monitoring.

I am using a classification proxy because I can mark pages as declining or not declining using the dataset column trend_direction.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

My target/proxy is whether a page is declining.

I define:

is_declining_label = 1 if trend_direction == "down"
is_declining_label = 0 otherwise

This label comes from a defined rule using the observed trend_direction column. It is a proxy because it does not prove that refreshing the page will recover traffic. It only shows that the page had a measured downward trend.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The main success metric is Precision@50.

This is useful because a content team cannot review every page. They may only review the top 50 pages recommended by the scoring system.

Precision@50 tells us how many of the top 50 recommended pages are actually declining.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis is one content item/page.

Each row represents one pseudonymized content page, identified by content_id. The row contains search visibility, clicks, sessions, position, age, freshness, and trend information.

The target column I created is is_declining_label.

```python
import pandas as pd
from pathlib import Path

data_dir = Path("data/raw")
data_dir.mkdir(parents=True, exist_ok=True)

data_path = data_dir / "content_refresh_anonymized.csv"

if not data_path.exists():
    url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
    df = pd.read_csv(url)
    df.to_csv(data_path, index=False)
else:
    df = pd.read_csv(data_path)

df.shape
```

```python
lane_cols = [
    "content_id",
    "client_id",
    "content_type",
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "clicks_90d",
    "sessions_90d",
    "ctr",
    "avg_position",
    "impression_tier",
    "position_tier",
    "trend_direction",
    "trend_pct",
]

lane_df = df[lane_cols].copy()

lane_df["is_declining_label"] = (
    lane_df["trend_direction"] == "down"
).astype(int)

lane_df.head(10)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule might say: review every page older than 180 days, or every page with low CTR.

That is too simple because page priority depends on many signals together: impressions, clicks, CTR, position, age, freshness, sessions, and content type.

A page can be old but still performing well. Another page can be newer but already declining. ML or scoring helps combine multiple signals and rank pages by likely priority.

The output supports a real content action: deciding which pages should be reviewed first for refresh, metadata improvement, expansion, protection, pruning, or monitoring.
